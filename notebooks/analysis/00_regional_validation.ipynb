Síntese da validação regional

```python
from pathlib import Path
import sys
import pandas as pd

sys.path.append("/code/scripts")

from analysis_config import ANALYSIS
from validation_summary_utils import (
    controlled_validation_rows,
    read_validation_workbook,
    seasonal_validation_rows,
    sheets_to_long,
    validation_comparison_plan,
    workbook_inventory
)
```

```python
# ALTERAR APENAS SE O EXCEL ESTIVER NOUTRO CAMINHO
workbook = Path(
    "/code/data/results/validacoes_cenarios_C1_C8.xlsx"
)

out_dir = ANALYSIS / "00_validation_regional"
out_dir.mkdir(parents=True, exist_ok=True)
output = out_dir / "validation_regional_synthesis.xlsx"

print("Input:", workbook)
print("Output:", output)
```

```python
tables = read_validation_workbook(workbook)
inventory = workbook_inventory(tables)
long_table = sheets_to_long(tables)
comparison_plan = validation_comparison_plan()
controlled_rows, diagnostics = controlled_validation_rows(tables)
seasonal_rows = seasonal_validation_rows(tables)

inventory
```

```python
with pd.ExcelWriter(output) as writer:
    inventory.to_excel(
        writer,
        sheet_name="Inventory",
        index=False
    )
    comparison_plan.to_excel(
        writer,
        sheet_name="Comparison_plan",
        index=False
    )
    controlled_rows.to_excel(
        writer,
        sheet_name="Controlled_rows",
        index=False
    )
    seasonal_rows.to_excel(
        writer,
        sheet_name="Seasonal_results",
        index=False
    )
    diagnostics.to_excel(
        writer,
        sheet_name="Diagnostics",
        index=False
    )
    long_table.to_excel(
        writer,
        sheet_name="All_results",
        index=False
    )

    for sheet, table in tables.items():
        table.to_excel(
            writer,
            sheet_name=str(sheet)[:31],
            index=False
        )

print("Síntese guardada em:", output)
```
