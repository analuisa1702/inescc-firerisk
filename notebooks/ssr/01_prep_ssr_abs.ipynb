Preparação do SSR absoluto

Executar uma vez para `centro` e outra para `extremadura`.

```python
from pathlib import Path
import sys

sys.path.append("/code/scripts")

from seasonal_utils import dsr_to_ssr_raster
from raster_utils import check_raster_alignment
```

```python
# ALTERAR APENAS ESTA VARIÁVEL

area = "centro"


years = range(2008, 2026)

raw_dir = Path("/code/data/raw/meteo/cems_fire")

reference = Path(
    f"/code/data/processed/{area}/"
    f"topo/derived/dem_{area}_clip_complete.tif"  
)

out_dir = Path(
    f"/code/data/processed/{area}/meteo/ssr"
)
out_dir.mkdir(parents=True, exist_ok=True)

# Deixar None para deteção automática.
# Preencher apenas se o nome da variável não for reconhecido.
variable = None

print("Área:", area)
print("Referência:", reference)
print("Output:", out_dir)
```

```python
all_files = sorted(raw_dir.rglob("*.nc"))

if not all_files:
    print("Não foram encontrados ficheiros NetCDF em:", raw_dir)
else:
    print("NetCDF encontrados:", len(all_files))
```

```python
for year in years:
    year_files = [
        path
        for path in all_files
        if str(year) in path.name
    ]

    # Um único NetCDF pode conter vários anos.
    files = year_files if year_files else all_files

    output = out_dir / f"ssr_abs_{year}.tif"

    dsr_to_ssr_raster(
        netcdf_files=files,
        year=year,
        reference=reference,
        output=output,
        variable=variable
    )
```

## Verificação

```python
for year in years:
    output = out_dir / f"ssr_abs_{year}.tif"

    check_raster_alignment(
        raster=str(output),
        template=str(reference)
    )
```
