Targets anuais de área ardida de verão

Cria um target anual para a regressão logística:

- `1` — célula ardida entre **15 de junho e 15 de setembro**;
- `0` — célula válida não ardida nesse período;
- `-1` — área excluída ou NoData.

```python
from pathlib import Path
import sys

sys.path.append("/code/scripts")

from seasonal_utils import create_yearly_burned_target
from raster_utils import check_raster_alignment
```

```python
# ALTERAR APENAS ESTA VARIÁVEL

area = "centro"


years = range(2008, 2026)
source = "effis"
date_field = "initialdat"

vector_dir = Path(
    f"/code/data/processed/{area}/"
    f"area_ardida/{source}/yearly_aoi"
)

reference = Path(
    f"/code/data/processed/{area}/"
    f"topo/derived/dem_{area}_clip_complete.tif"  
)

out_dir = Path(
    f"/code/data/processed/{area}/"
    f"area_ardida/{source}/raster_target_yearly"
)
out_dir.mkdir(parents=True, exist_ok=True)

print("Área:", area)
print("Vetores anuais:", vector_dir)
print("Referência:", reference)
print("Output:", out_dir)
```

```python
for year in years:
    vector = vector_dir / f"aa_{year}.shp"
    output = out_dir / f"rst_ba_{year}_target.tif"

    create_yearly_burned_target(
        vector=vector,
        reference=reference,
        output=output,
        year=year,
        date_field=date_field
    )
```

## Verificação

```python
for year in years:
    output = out_dir / f"rst_ba_{year}_target.tif"

    check_raster_alignment(
        raster=str(output),
        template=str(reference)
    )
```
