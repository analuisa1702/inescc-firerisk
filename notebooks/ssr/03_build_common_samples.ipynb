Amostra comum para C7 e C8

Constrói uma única amostra espacial para os dois cenários:

- **400 células por ano**;
- **200 ardidas** e **200 não ardidas**;
- treino: **2008–2024**;
- validação independente: **2025**.

```python
from pathlib import Path
import sys

sys.path.append("/code/scripts")

from seasonal_utils import build_common_samples
```

```python
# ALTERAR APENAS ESTA VARIÁVEL

area = "centro"


train_years = range(2008, 2025)
validation_year = 2025

n_burned = 200
n_unburned = 200
seed = 42

susc_lr = Path(
    f"/code/data/results/{area}/C5/final/res_lri.tif"
)

susc_rf = Path(
    f"/code/data/results/{area}/C6/lri_model/class/"
    "rf_lri_base_mean_prob_1_10models.tif"
)

target_dir = Path(
    f"/code/data/processed/{area}/area_ardida/effis/"
    "raster_target_yearly"
)

ssr_dir = Path(
    f"/code/data/processed/{area}/meteo/ssr"
)

out_dir = Path(
    f"/code/data/processed/{area}/model_inputs/seasonal"
)
out_dir.mkdir(parents=True, exist_ok=True)

train_csv = out_dir / "samples_train_2008_2024.csv"
valid_csv = out_dir / "samples_valid_2025.csv"

print("Área:", area)
print("Suscetibilidade LR:", susc_lr)
print("Suscetibilidade RF:", susc_rf)
print("Amostras:", out_dir)
```

```python
train = build_common_samples(
    years=train_years,
    target_folder=target_dir,
    ssr_folder=ssr_dir,
    susceptibility_lr=susc_lr,
    susceptibility_rf=susc_rf,
    output_csv=train_csv,
    n_burned=n_burned,
    n_unburned=n_unburned,
    seed=seed
)
```

```python
valid = build_common_samples(
    years=[validation_year],
    target_folder=target_dir,
    ssr_folder=ssr_dir,
    susceptibility_lr=susc_lr,
    susceptibility_rf=susc_rf,
    output_csv=valid_csv,
    n_burned=n_burned,
    n_unburned=n_unburned,
    seed=seed
)
```

## Verificação da distribuição anual

```python
train_counts = train.groupby(
    ["year", "burned"]
).size().unstack(fill_value=0)

valid_counts = valid.groupby(
    ["year", "burned"]
).size().unstack(fill_value=0)

print("Treino:")
display(train_counts)

print("Validação:")
display(valid_counts)

train.head()
```
