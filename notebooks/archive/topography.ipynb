### Cálculo perigosidade Incêndio Florestal

Gerar e reclassificar variáveis topográficas:

- Elevation
- Slope
- Aspect

```python
dem = '/home/jasp/bedsburning/dem_ctr.tif'

ws = '/home/jasp/bedsburning'

bname = 'ctr_lri'
```

### Python packages

```python
import os

from glass.ete.lri import HeuristicLri
```

### Workspace

```python
lri = HeuristicLri(ws, dem, loc=None)
```

### Generate Slope and Aspect

```python
lri.gen_slope_aspect(
    rst_slope=os.path.join(ws, f'{bname}_slope.tif'),
    rst_aspect=os.path.join(ws, f'{bname}_aspect.tif')
)
```

### Reclassify DEM, Slope and Aspect

```python
lri.recls_topo(
    rcls_dem=os.path.join(ws, f'{bname}_var_dem.tif'),
    rcls_slope=os.path.join(ws, f'{bname}_var_slope.tif'),
    rcls_aspect=os.path.join(ws, f'{bname}_var_aspect.tif')
)
```
