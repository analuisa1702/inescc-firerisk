# Preparação dos Rasters de LULC
* Conversão das camadas harmonizadas de uso e ocupação do solo para formato raster.  

```python
from pathlib import Path
from glob import glob
import rasterio as rio
import numpy as np

from glass.dtt.rst.torst import shp_to_rst
```

```python
fld = "/code/data/processed/pnse/lulc"
ref = "/code/data/processed/pnse/topo/derived/dem_pnse.tif"

in_dir = f"{fld}/harmonized_vectors"
out_dir = f"{fld}/rasters"

Path(out_dir).mkdir(parents=True, exist_ok=True)
```

```python
lulc_files = {
    1995: {
        "file": f"{in_dir}/cos_1995_harm.gpkg",
        "layer": "cos_1995_harm",
        "out": f"{out_dir}/lulc_1995.tif",
    },
    2007: {
        "file": f"{in_dir}/cos_2007_harm.gpkg",
        "layer": "cos_2007_harm",
        "out": f"{out_dir}/lulc_2007.tif",
    },
    2010: {
        "file": f"{in_dir}/cos_2010_harm.gpkg",
        "layer": "cos_2010_harm",
        "out": f"{out_dir}/lulc_2010.tif",
    },
    2015: {
        "file": f"{in_dir}/cos_2015_harm.gpkg",
        "layer": "cos_2015_harm",
        "out": f"{out_dir}/lulc_2015.tif",
    },
    2018: {
        "file": f"{in_dir}/cos_2018_harm.gpkg",
        "layer": "cos_2018_harm",
        "out": f"{out_dir}/lulc_2018.tif",
    },
}
```

```python
#rasterizr tudo com o id_harm
for year, cfg in lulc_files.items():
    shp_to_rst(
        shp=cfg["file"],
        inSource="id_harm",
        cellsize=None,
        nodata=0,
        outRaster=cfg["out"],
        rst_template=ref,
        lyrname=cfg["layer"],
        api="gdal",
        dtype="UInt16",
        rtype=int,
    )

    print(f"{year} -> {cfg['out']}")
```

```python
#verificação
for year, cfg in lulc_files.items():
    with rio.open(cfg["out"]) as src:
        arr = src.read(1)
        vals = np.unique(arr)
        vals = vals[vals != 0]

        print(f"\nAno: {year}")
        print("n classes presentes:", len(vals))
        print("ids presentes:", vals[:50])
```
