### Preparação de variáveis para cálculo do LRI

**Variáveis:**

- Elevation
- Slope

```python
area = "extremadura"
# Opções: "centro" ou "extremadura"

base = f"/code/data/processed/{area}/topo"

dem = f"{base}/derived/dem_{area}_clip.tif"

out_slope = f"{base}/derived/slope_{area}.tif"
out_aspect = f"{base}/derived/aspect_{area}.tif"

rcls_dem = f"{base}/reclassified/rcls_dem_{area}.tif"
rcls_slope = f"{base}/reclassified/rcls_slope_{area}.tif"
rcls_aspect = f"{base}/reclassified/rcls_aspect_{area}.tif"

slope = None
aspect = None
```

```python
import os
from glass.ete.lri import HeuristicLri
```

```python
# Start Class
lri = HeuristicLri(os.path.dirname(dem), dem, loc=None)

# Import topographic variables

lri.import_topo_vars(dem=dem, slope=slope, aspect=aspect)
```

```python
# Generate Slope and Aspect if necessary
if not slope or not aspect:
    lri.gen_slope_aspect(
        rst_slope=out_slope,
        rst_aspect=out_aspect
    )
```

```python
# Reclassify Topographic Variables

lri.recls_topo(
    rcls_dem=rcls_dem,
    rcls_slope=rcls_slope,
    rcls_aspect=rcls_aspect
)
```
