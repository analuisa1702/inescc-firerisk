```python
refs = {
    '1995-2006' : '/mnt/x/susmod_tk2/burn_areas/burned_1995_2006.tif',
    '2007-2009' : '/mnt/x/susmod_tk2/burn_areas/burned_2007_2009.tif',
    '2010-2014' : '/mnt/x/susmod_tk2/burn_areas/burned_2010_2014.tif',
    '2015-2017' : '/mnt/x/susmod_tk2/burn_areas/burned_2015_2017.tif',
    '2022'      : '/mnt/x/susmod_tk2/burn_areas/burned_2022.tif'
}

perigos = [
    '/mnt/x/susmod_tk2/lri_results/sus_1995_2006.tif',
    '/mnt/x/susmod_tk2/lri_results/sus_2007_2009.tif',
    '/mnt/x/susmod_tk2/lri_results/sus_2010_2014.tif',
    '/mnt/x/susmod_tk2/lri_results/sus_2015_2017.tif'
]

val = 1

otbl = '/mnt/x/susmod_tk2/lri_results/sintese.xlsx'
```

```python
import os
import pandas as pd
from glass.pys.oss import fprop
from glass.ete.perig.eval import pseudo_roc
from glass.wt import obj_to_tbl
```

```python
ftbl = []

reflst = list(refs.keys())
for p in perigos:
    pname = fprop(p, "fn")
    row = [pname]
    for ref in reflst:
        vtbl, rauc, pseudo_auc = pseudo_roc(refs[ref], p, val, os.path.join(os.path.dirname(otbl), f'{pname}_{ref}.xlsx'))

        row.append(rauc)

    ftbl.append(row)

fdf = pd.DataFrame(ftbl, columns=['perigosity'] + reflst)

obj_to_tbl(fdf, otbl)
```
