### Produce training and validation samples

```python
# ALTERAR APENAS ESTAS DUAS VARIÁVEIS

scenario = "C6"
area = "extremadura"

# C2 e C4: area = "centro"
# C6: area = "centro" ou "extremadura"


# C2 — ICNF | 1995–2024
if scenario == "C2":
    train_period = "1995_2024"


# C4 e C6 — ICNF ou EFFIS | 2008–2024
elif scenario in ["C4", "C6"]:
    train_period = "2008_2024"


target_folder = (
    f"/code/data/results/{area}/"
    f"{scenario}/lri_model/target"
)

train_folder = (
    f"/code/data/results/{area}/"
    f"{scenario}/lri_model"
)


fires = {
    f"f{train_period}": (
        f"{target_folder}/"
        f"rst_ba_{train_period}_target.tif"
    )
}


# Amostragem: 10 000 presenças e 50 000 pseudo-ausências
class_abs = {
    1: 10000,
    0: 50000,
}


# Dez réplicas independentes
nsets = 10


print("Cenário:", scenario)
print("Área:", area)
print("Período:", train_period.replace("_", "–"))
print("Target:", list(fires.values())[0])
print("Número de réplicas:", nsets)
```

```python
import os
import numpy as np
import sys
from glass.pys.oss import mkdir

sys.path.append("/code/scripts")
from sample import split_rst_radomly_byblocks
```

```python
for k in fires:
    trn_fld = mkdir(os.path.join(train_folder, f'train_{k}'))

    for i in range(1, nsets + 1):
        np.random.seed(i)

        split_rst_radomly_byblocks(
            fires[k],
            class_abs,
            os.path.join(trn_fld, f'{k}_{str(i)}.tif'),
            other_rst=None,
            absbycls=True,
            block_size=1024
        )
```
