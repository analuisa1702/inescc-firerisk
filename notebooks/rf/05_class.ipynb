```python
xlsx = f'/code/data/results/{area}/{scenario}/lri_model/models.xlsx'
sheet = 'models'

import os
#from glass.rst.cls import imgcls_from_mdl
from glass.rd import tbl_to_obj
import datetime as dt


import sys
sys.path.append('/code/scripts')

from sample import rf_mean_prob_byblocks

df = tbl_to_obj(xlsx, sheet)
```

```python
df = df[df['status'] == 'run']

models = df['model'].tolist()

featdf = tbl_to_obj(xlsx, df.iloc[0]['name'])

feats = featdf.predfeat.tolist()
paths = [
    os.path.join(df.iloc[0]['predfeat'], f)
    for f in feats
]

out_mean = (
    f'/code/data/results/{area}/{scenario}/lri_model/class/'
    'rf_lri_base_mean_prob_1_10models.tif'
)

print('N.º de modelos:', len(models))
print('Variáveis:', paths)
print('Output:', out_mean)
```
