### Produce training and validation samples

```python
fires = {
    #'f95' : '/mnt/x/susmod_tk2/burn_areas/burned_1995_2006_bin.tif',
    #'f07' : '/mnt/x/susmod_tk2/burn_areas/burned_2007_2009_bin.tif',
    #'f10' : '/mnt/x/susmod_tk2/burn_areas/burned_2010_2014_bin.tif',
    'f15' : '/mnt/x/susmod_tk2/burn_areas/burned_2015_2017_bin.tif'
}

class_abs = {1 : 10000, 0 : 50000}

train_folder = '/mnt/x/susmod_tk2/rf_modl1/'
test__folder = '/mnt/x/susmod_tk2/rf_modl1/'

nsets = 10
```

```python
import os
from glass.pys.oss import mkdir
from glass.smp import split_rst_radomly
```

```python
for k in fires:
    tst_fld = mkdir(os.path.join(test__folder, f'tst_{k}'))
    trn_fld = mkdir(os.path.join(train_folder, f'train_{k}'))

    for i in range(1, nsets+1):
        split_rst_radomly(
            fires[k], class_abs,
            os.path.join(trn_fld, f'{k}_{str(i)}.tif'),
            os.path.join(tst_fld, f'{k}_{str(i)}.tif'),
            absbycls=True
        )
```
