```python
import os
import pandas as pd
```

```python
area = "centro"

rf_folder = f"/code/data/results/{area}/rf/lri_model"
lr_final_folder = f"/code/data/results/{area}/lr/final"

train_folder = os.path.join(rf_folder, "train_f1995_2024")

xlsx = os.path.join(rf_folder, "models.xlsx")

os.makedirs(rf_folder, exist_ok=True)
os.makedirs(os.path.join(rf_folder, "npy"), exist_ok=True)
os.makedirs(os.path.join(rf_folder, "models"), exist_ok=True)
os.makedirs(os.path.join(rf_folder, "class"), exist_ok=True)

print("Excel será criado em:", xlsx)
print("Pasta dos rasters de treino:", train_folder)
print("Pasta das variáveis LRI:", lr_final_folder)
```

```python
#definir os modelos
model_features = {
    "rf_lri_base": [
        "lri_dem.tif",
        "lri_slope.tif",
        "lri_lulc.tif"
        #"lri_aspect.tif"
    ]
}
```

```python
#criar a folha models
nsets = 10 #alterar para os 50 modelos

rows = []

for model_name in model_features:
    for i in range(1, nsets + 1):
        run_name = f"{model_name}_{i}"

        trainref = os.path.join(
            train_folder,
            f"f1995_2024_{i}.tif"
        )

        result_folder = os.path.join(rf_folder, "class", run_name)
        os.makedirs(result_folder, exist_ok=True)

        rows.append({
            "status": "run",
            "name": model_name,
            "trainref": trainref,
            "trainfeat": lr_final_folder,
            "predfeat": lr_final_folder,
            "yfile": os.path.join(rf_folder, "npy", f"{run_name}_y.npy"),
            "xfile": os.path.join(rf_folder, "npy", f"{run_name}_x.npy"),
            "model": os.path.join(rf_folder, "models", f"{run_name}.joblib"),
            "ntrees": 200, 
            "max_samples": -1, 
            "result": os.path.join(result_folder, f"{run_name}.tif")
        })

df_models = pd.DataFrame(rows)

df_models
```

```python
#verficação
missing_train = [
    p for p in df_models["trainref"].tolist()
    if not os.path.exists(p)
]

if missing_train:
    raise FileNotFoundError(
        f"Faltam {len(missing_train)} rasters de treino. Exemplo: {missing_train[0]}"
    )

for model_name, feats in model_features.items():
    for f in feats:
        fpath = os.path.join(lr_final_folder, f)

        if not os.path.exists(fpath):
            raise FileNotFoundError(f"Falta variável LRI: {fpath}")

print("Todos os rasters de treino e variáveis LRI existem.")
```

```python
#criar o excel
with pd.ExcelWriter(xlsx) as writer:
    df_models.to_excel(
        writer,
        sheet_name="models",
        index=False
    )

    for model_name, feats in model_features.items():
        df_feats = pd.DataFrame({
            "trainfeat": feats,
            "predfeat": feats
        })

        df_feats.to_excel(
            writer,
            sheet_name=model_name,
            index=False
        )

print("models.xlsx criado:", xlsx)
```
