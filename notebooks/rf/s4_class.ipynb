```python
xlsx = '/code/data/results/centro/rf/lri_model/models.xlsx'
sheet = 'models'

import os
from glass.rst.cls import imgcls_from_mdl
from glass.rd import tbl_to_obj
import datetime as dt

df = tbl_to_obj(xlsx, sheet)
```

```python
for i, r in df.iterrows():
    if r['status'] != 'run':
        continue
        
    time_a = dt.datetime.now().replace(microsecond=0)
    featdf = tbl_to_obj(xlsx, r['name'])

    feats = featdf.predfeat.tolist()
    paths = [os.path.join(r.predfeat, f) for f in feats]
    
    imgcls_from_mdl(
        r.model, paths, r['result'],
        probrst=os.path.dirname(r['result'])
    )
    time_b = dt.datetime.now().replace(microsecond=0)

    print(time_b - time_a)
```
