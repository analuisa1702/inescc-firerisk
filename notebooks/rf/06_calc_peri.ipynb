# Cálculo da perigosidade 
* Suscetibilidade x Probabilidade

```python
import os
import sys

sys.path.append("/code/scripts")

from lri_sum import HeuristicLriCount
```

```python
# ALTERAR APENAS ESTAS DUAS VARIÁVEIS

area = "extremadura"
scenario_rf = "C6"


# Associação entre o cenário RF e o respetivo cenário LR

if scenario_rf == "C2":
    scenario_lr = "C1"

elif scenario_rf == "C4":
    scenario_lr = "C3"

elif scenario_rf == "C6":
    scenario_lr = "C5"


dem = (
    f"/code/data/processed/{area}/"
    f"topo/reclassified/rcls_dem_{area}.tif"
)


out_dir_rf = (
    f"/code/data/results/{area}/"
    f"{scenario_rf}/lri_model/class"
)

out_dir_lr = (
    f"/code/data/results/{area}/"
    f"{scenario_lr}/final"
)


os.makedirs(
    out_dir_rf,
    exist_ok=True
)


ws = out_dir_rf


# Suscetibilidade média produzida pelos 10 modelos RF
out_susc = (
    f"{out_dir_rf}/"
    "rf_lri_base_mean_prob_1_10models.tif"
)


# Probabilidade histórica calculada no cenário LR correspondente
out_prob = (
    f"{out_dir_lr}/"
    "wprobability.tif"
)


# Perigosidade RF
out_peri = (
    f"{out_dir_rf}/"
    "res_perigosity_rf.tif"
)


print("Cenário RF:", scenario_rf)
print("Área:", area)
print("Cenário LR correspondente:", scenario_lr)
print("Suscetibilidade RF:", out_susc)
print("Probabilidade:", out_prob)
print("Perigosidade:", out_peri)
```

```python
#inicializar lri
from lri_sum import HeuristicLriCount
lri = HeuristicLriCount(ws, dem, loc=None)
```

```python
# Generate perigosity 
lri.get_perigosity(
    inlri=out_susc,
    wildprob=out_prob,
    out=out_peri
)

print(f"Perigosidade RF criada em: {out_peri}")
```
