```python
refs = {
    '2015-2023' : '/mnt/x/firebeds_tk1/aa/rst_aa_all.tif',
    '2024'      : '/mnt/x/firebeds_tk1/aa_rst/rst_ardida_2024.tif'
}

perigos = {
    'dem'    : '/mnt/x/firebeds_tk1/results2/lri_dem.tif',
    'slope'  : '/mnt/x/firebeds_tk1/results2/lri_slope.tif',
    'aspect' : '/mnt/x/firebeds_tk1/results2/lri_aspect.tif',
    'lulc'   : '/mnt/x/firebeds_tk1/results2/lri_lulc.tif',
    'lri'    : '/mnt/x/firebeds_tk1/results2/res_lri.tif',
    'prob'   : '/mnt/x/firebeds_tk1/results2/wprobability.tif',
    'peri'   : '/mnt/x/firebeds_tk1/results2/res_perigosity.tif'
}

val = 1

otbl = '/mnt/x/firebeds_tk1/results2/valida_lri.xlsx'
```

```python
import os
import pandas as pd
from glass.pys.oss import fprop, lst_ff
from glass.ete.perig.eval import pseudo_roc
from glass.wt import obj_to_tbl
```

```python
ftbl = []

reflst = list(refs.keys())
for p in perigos:
    rname = fprop(perigos[p], "fn")
    row = [p, rname]
    for ref in reflst:
        vtbl, rauc, pseudo_auc = pseudo_roc(
            refs[ref], perigos[p], val,
            os.path.join(os.path.dirname(otbl), f'{p}_{rname}_{ref}.xlsx')
        )

        row.append(rauc)

    ftbl.append(row)

fdf = pd.DataFrame(ftbl, columns=['alias', 'perigosity'] + reflst)

obj_to_tbl(fdf, otbl)
```
